```python
import torch
import numpy as np
import pandas as p
```

```python
# 텐서 생성
# 0-way Tensor
scalar = torch.tensor(5)
print(scalar)
print(scalar.shape)  #tensor 안에 ndarray 가 들어가있으니까 shape을 쓸수있다
```

```python

# 1차원   vector는 여러개값이 들어가있음 - 단일값이 아님
vector = torch.tensor([1, 2, 3, 4])
print(vector)
print(vector.shape)
```

```python
# 2차원
matrix = torch.tensor([[1, 2, 3], [4, 5, 6]])
print(matrix)
print(matrix.shape)
```

```python
# 3차원
tensor3d = torch.rand(2, 3, 4) #random하게 2x3x4 텐서 생성   .rand
print(tensor3d)
print(tensor3d.shape)
```

```python
#텐서 다루기 -> 텐서의 타입 지정
int_tensor = torch.tensor([1, 2, 3], dtype=torch.int64)
float_tensor = torch.tensor([1.0, 2.0, 3.0], dtype=torch.float32)
print(int_tensor)
print(float_tensor)
```

```python
# 2.2 텐서 속성 확인
print(float_tensor)
print(f'Shape: {float_tensor.shape}')
print(f'Datatype: {float_tensor.dtype}')
print(f'Device: {float_tensor.device}')
```

```python
# 2.3 텐서 연산하기
a = torch.tensor([1.0, 2.0, 3.0])
b = torch.tensor([4.0, 5.0, 6.0])
print('덧셈:', a + b)
print('곱셈:', a * b)
print('제곱:', a ** 2)
```

```python
x = torch.tensor([[1, 2, 3], [4, 5, 6]], dtype=torch.float32)
print('전체 합계:', x.sum())
print('열별 합계:', x.sum(dim=0))
print('행별 합계:', x.sum(dim=1))
```

```python
# 2.5 텐서 차원 변경
x = torch.tensor([[1, 2], [3, 4], [5, 6]])
print('원래 텐서:', x)
reshaped = x.view(2, 3)
print('변형된 텐서:', reshaped)
```

```python
# 2.6 텐서 연결과 분리
a = torch.tensor([[1, 2], [3, 4]])
b = torch.tensor([[5, 6]])


# 텐서 연결하기
concatenated = torch.cat([a, b], dim=0)
print('연결된 텐서:\n', concatenated)
```

```python
# 텐서 분리하기
x = torch.tensor([1, 2, 3, 4, 5, 6])
split1, split2 = torch.split(x, 3)
print('분리된 텐서:', split1, split2)
```

```python
# 2.7 텐서 내적과 형렬곱
# 내적
a = torch.tensor([1, 2, 3])
b = torch.tensor([4, 5, 6])
dot = torch.dot(a, b)
print('Dot product:', dot)
```

```python
# 행렬곱
A = torch.tensor([[1, 2], [3, 4]])
B = torch.tensor([[5, 6], [7, 8]])


result = torch.matmul(A, B)
print(result)
```

```python
import torch
import numpy as np
import pandas as pd



# 광고비에 따른 매출액 예측
# DataLoader
path = 'https://bit.ly/advertising_csv'
adv = pd.read_csv(path)
adv.head()
```

```python
adv.info()
```

```python
target = 'Sales'  #원래답
x = adv.drop(target, axis=1) #원래답 제거 ->  데이터만 남곘다

y = adv.loc[:, target]
x_np = x.values
y_np = y.values
print(type(x_np))
print(x_np.shape)
print(type(y_np))
print(y_np.shape)
```

```python

# 넘파이 배열 x, y를 텐서로 변환
x_tensor = torch.tensor(x_np, dtype=torch.float32)
y_tensor = torch.tensor(y_np, dtype=torch.float32)
print(x_tensor.shape, y_tensor.shape)
from torch.utils.data import DataLoader, TensorDataset
dataset = TensorDataset(x_tensor, y_tensor)
print(dataset[:3])
print(len(dataset))
```

```python
# 데이터 로더 생성
data_loader = DataLoader(dataset, batch_size = 32, shuffle = True)

for n, batch in enumerate(data_loader):
    x_batch, y_batch = batch
    print(f'배치 {n}, x 배치 크기:{x_batch.shape}')

```

```python
# 디바이스 할당
# 1. 데이터 준비
# 라이브러리 로딩
import torch
from torch import nn
import numpy as np
import pandas as pd


# 데이터 로딩 및 x, y 분할
path = 'https://bit.ly/advertising_csv'
adv = pd.read_csv(path)


target = 'Sales'
x = adv.drop(target, axis=1)
y = adv.loc[:, target]


# numpy로 변환한 다음 텐서로 변환
x_np = x.values
y_np = y.values
x_tensor = torch.tensor(x_np, dtype=torch.float32)
y_tensor = torch.tensor(y_np, dtype=torch.float32)
```

```python
# 디바이스 준비
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'Using {device} device.')
```

```python
#모델과 데이터셋을 디바이스 할당
#모델을 디바이스에 할당 (GPU)
model = nn.Sequential( nn.Linear(3, 1) )
model = model.to(device)
```

```python
# 데이터 세트를 디바이스에 할당.
x, y = x_tensor.to(device), y_tensor.to(device)
```

```python
# gpu -> cpu로 이동x
x_cpu = x.cpu()

y_cpu = y.cpu()
print(x_cpu.device, y_cpu.device)
```
