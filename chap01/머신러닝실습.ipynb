```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.model_selection import train_test_split
#from sklearn.metrics import *
from sklearn.metrics import root_mean_squared_error
from sklearn.metrics import r2_score
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_absolute_percentage_error

from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

```python
#1. 비즈니스 이해  2.데이터 이해 3.데이터 전처리 4.모델링 5.평가 6.배포
# 3.데이터 전처리
# 3.1 데이터 준비: http://bit.ly/advertising.csv
path = 'http://bit.ly/advertising_csv'
data1 = pd.read_csv(path)
data1.head()
```

```python
target = 'Sales'
X = data1.drop(target, axis=1)  #remove sales columns
y =data1.loc[:,target] #set sales columns
```

```python
#data splitting
#train_test_split() 함수는 데이터를 학습용과 테스트용으로 나누어주는 함수입니다.
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=20)
```

```python
scaler = MinMaxScaler()
# scaler.fit(X_train)
# X_train_scaled = scaler.transform(X_train) #proetect original X_train data

#학습데이터에 대한 변환
X_train_scaled = scaler.fit_transform(X_train) #fit and transform X_train data
#테스트데이터에 대한 변환
X_test_scaled = scaler.transform(X_test) #transform X_test data
```

```python
#modeling   fit() -> predict() -> evaluate()
#model selection: Linear regression 
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train_scaled, y_train)

#pred
preds = model.predict(X_test_scaled)  #pred  for test data  <- test data is new data
```

```python
#eval
print(f'RMSE: {root_mean_squared_error(y_test, preds)}') #
print(f'R2: {r2_score(y_test, preds)}')
print(f'MAE: {mean_absolute_error(y_test, preds)}')
```
